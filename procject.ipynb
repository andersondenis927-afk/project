## Data Objects and Attribute Types

#### This dataset contains only nominal and numeric attributes and this is intuitively clear when looking at this dataset :
- Nominal attribute:
    - Day — represents dates, which serve as identifiers rather than ordered categories.
- Numeric attributes:
    - Weight_kg
    - Calories_Intake
    - Protein_Intake_g
    - Workout_Duration_min
    - Steps_Walked

* These are continuous variables suitable for mathematical operations and statistical analysis.
* There are no binary attributes (columns with exactly two unique values) and no ordinal attributes (categorical variables with a meaningful order, such as size levels).
